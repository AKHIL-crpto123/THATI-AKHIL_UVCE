```python
import speech_recognition as sr

recognizer = sr.Recognizer()

with sr.AudioFile('/harvard.wav') as source:
    audio_data = recognizer.record(source)

# Convert audio to text using Google's API
text = recognizer.recognize_google(audio_data)

# Print the transcribed text
print(text)
```

```python
import language_tool_python

tool = language_tool_python.LanguageTool('en-US')
matches = tool.check(text)
grammar_score = len(matches)
print(f"Grammar Score: {grammar_score}")
```

```python
import matplotlib.pyplot as plt

# Example grammar scores
grammar_score_1 = 5
grammar_score_2 = 3
grammar_score_3 = 7

scores = [grammar_score_1, grammar_score_2, grammar_score_3]
plt.plot(scores, marker='o')
plt.title('Grammar Scores')
plt.xlabel('Sample Number')
plt.ylabel('Number of Grammar Issues')
plt.grid(True)
plt.show()
```
